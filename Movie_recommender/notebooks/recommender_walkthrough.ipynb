# Code-A-Nova Machine Learning Internship
## Task: Level 2(b) - Movie Recommendation System

### Project Objective
Build an end-to-end movie recommendation engine using the official **MovieLens** dataset that recommends films using:
1. **Content-Based Filtering**: TF-IDF Vectorization & Cosine Similarity on movie genre metadata.
2. **Collaborative Filtering**: Matrix Factorization using Truncated SVD on the User-Item interaction matrix.
3. **Hybrid Recommender**: Blends content similarity with user taste profiles.

```python
import os
import sys
from pathlib import Path

# Ensure project root is in sys.path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

import pandas as pd
import numpy as np
from src.preprocess import load_movielens_data, clean_movie_metadata, compute_movie_stats
from src.content_based import ContentBasedRecommender
from src.collaborative import CollaborativeRecommender
from src.engine import MovieRecommenderEngine

print("All modules loaded successfully!")
```

### 1. Load and Inspect the MovieLens Dataset

```python
movies, ratings = load_movielens_data()
movies_clean = clean_movie_metadata(movies)
movies_enriched = compute_movie_stats(movies_clean, ratings)
movies_enriched.head()
```

### 2. Content-Based Filtering (Cosine Similarity)

```python
content_rec = ContentBasedRecommender(movies_enriched)

# Recommend movies similar to 'Toy Story'
recs = content_rec.recommend('Toy Story', top_n=5)
pd.DataFrame(recs)
```

### 3. Collaborative Filtering (Matrix Factorization via SVD)

```python
collab_rec = CollaborativeRecommender(ratings, movies_enriched, n_components=25)

# Personalized recommendations for User #1
user_recs = collab_rec.recommend_for_user(user_id=1, top_n=5)
pd.DataFrame(user_recs)
```

### 4. Hybrid Recommendation Engine

```python
engine = MovieRecommenderEngine()
hybrid_results = engine.hybrid_recommend(user_id=1, context_movie_title='The Matrix', top_n=5)
pd.DataFrame(hybrid_results)
```
